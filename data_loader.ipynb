```python
import matplotlib.pyplot as plt
import numpy as np
fin = int(5 * record.fs)
seg = record.p_signal[0:fin, 0]
plt.plot(seg)
plt.title('ECG Signal')
plt.xlabel('Samples')
plt.ylabel('Amplitude')
plt.show()
#Ahora, deseamos que el eje "x" esté en segundos, por lo que debemos dividir el número de muestras entre la frecuencia de muestreo. 
#Esto se puede hacer de la siguiente manera:
time = np.arange(0, fin) / record.fs
plt.plot(time, seg)
picos = annotation.sample[annotation.sample < fin]
timehigh = picos / record.fs
plt.scatter(timehigh, seg[picos], color='red', label='R-peaks')
plt.title('ECG Signal with R-peaks')
plt.show()
```
